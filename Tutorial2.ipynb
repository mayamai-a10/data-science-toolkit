```python
import numpy as np

array_2x2 = np.array ([
    [1,2],
    [3,4]
])

print("This is my numpy 2x2 array")
print(array_2x2)
```

```python
random3x4 = np.random.rand(3,4)

print("Random 3x4 array")
print(random3x4)
```

```python
ones_array = np.ones((3,3))
zeros_array = np.zeros((3,3))

print("Array of Ones")
print(ones_array)

print("Array of Zeroes")
print(zeros_array)
```

```python
random6x3 = np.random.rand(6,3)
transpose_array = random6x3.T

print("Random 6x3 Array")
print(random6x3)

print("Transpose 3x6 Array")
print(transpose_array)
```

```python
n = 3
p = 4
m = 2

array1 = np.random.rand(n,p)
array2 = np.random.rand(p,m)

dot_product = np.dot(array1,array2)

print(dot_product)
```
